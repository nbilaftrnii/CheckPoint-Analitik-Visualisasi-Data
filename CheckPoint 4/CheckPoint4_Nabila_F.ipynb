### **Outliers Values**

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

display(results_df)
```

## **Comparison/Perbandingan**

Bar Chart (Grafik Batang).

```python
sns.set_style("whitegrid")

df['age_group'] = pd.cut(df['age'],
                         bins=[18, 30, 45, 60, 100],
                         labels=['18-30', '31-45', '46-60', '60+'])

factors = ['age_group', 'sex', 'children', 'region', 'smoker']
ranking = []

for col in factors:
    mean_values = df.groupby(col, observed=False)['charges'].mean()

    ranking.append({"factor": col.replace("_"," ").title(),
                    "avg_charges": mean_values.max()})

ranking_df = pd.DataFrame(ranking).sort_values(by="avg_charges",
                                               ascending=True)

plt.figure(figsize=(12,7))

sns.barplot(data=ranking_df,
            y="factor",
            x="avg_charges",
            hue="factor",
            palette="viridis",
            legend=False)

plt.title("Ranking Faktor dengan Kontribusi Biaya Asuransi Tertinggi", fontsize=16)
plt.xlabel("Rata-rata Charges")
plt.ylabel("Faktor")

plt.tight_layout()
plt.show()
```

Hasil perbandingan menunjukkan bahwa status merokok memiliki rata-rata biaya asuransi tertinggi dibandingkan faktor lainnya.
```python
df['age_group'] = pd.cut(df['age'],
                         bins=[18, 30, 45, 60, 100],
                         labels=['18-30', '31-45', '46-60', '60+'])

factors = ['age_group', 'sex', 'children', 'region', 'smoker']

composition_list = []

for col in factors:
    temp = df[col].value_counts().reset_index()
    temp.columns = ['category', 'count']
    temp['factor'] = col
    temp['label'] = col + ": " + temp['category'].astype(str)
    composition_list.append(temp)

composition_df = pd.concat(composition_list)


threshold = composition_df['count'].sum() * 0.03

small = composition_df[composition_df['count'] < threshold]['count'].sum()
composition_df = composition_df[composition_df['count'] >= threshold]

if small > 0:
    composition_df = pd.concat([composition_df,
                                pd.DataFrame({'label':['Others'],
                                              'count':[small]})])

palette = sns.color_palette("tab10", len(composition_df))

plt.figure(figsize=(10,10))

wedges, texts, autotexts = plt.pie(composition_df['count'],
                                   labels=composition_df['label'],
                                   autopct='%1.1f%%',
                                   startangle=90,
                                   colors=plt.cm.Paired.colors,
                                   pctdistance=0.85)

plt.title('Komposisi Data Berdasarkan Berbagai Faktor', fontsize=16)
plt.tight_layout()
plt.show()
```

Grafik pie di atas menunjukkan bahwa beberapa kategori memiliki proporsi lebih dominan dibandingkan kategori lainnya, seperti kelompok usia tertentu dan kelompok non-perokok. Komposisi kategori yang beragam menunjukkan bahwa dataset mencerminkan berbagai karakteristik populasi yang berbeda.

# **Construct Data**

Tahap data construction dilakukan untuk membentuk variabel baru yang lebih informatif dan relevan berdasarkan data yang sudah ada, sehingga dapat meningkatkan kualitas analisis dan pemodelan.

1. **Membuat kolom baru "bmi category"**

    Pada tahap data construction, dilakukan pembentukan variabel baru yaitu BMI Category yang berasal dari variabel bmi. Pengelompokan dilakukan menggunakan fungsi kustom (categorize_bmi) dengan pendekatan kondisional (if-elif-else) berdasarkan standar World Health Organization (WHO).

    Variabel ini bertujuan untuk mengelompokkan tingkat risiko kesehatan individu sehingga mempermudah analisis hubungan antara kondisi tubuh dengan biaya asuransi, tanpa menambah kompleksitas data.

```python
def categorize_bmi(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25:
        return 'Normal'
    elif bmi < 30:
        return 'Overweight'
    else:
        return 'Obese'

df['bmi category'] = df['bmi'].apply(categorize_bmi)
```
