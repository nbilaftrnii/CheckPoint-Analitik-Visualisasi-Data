### **Outliers Values**

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

display(results_df)
```

# **Eksplorasi Data (EDA)**

## **Comparison/Perbandingan**

**Aktivitas:**

Membandingkan rata-rata biaya asuransi (charges) berdasarkan berbagai faktor yang mempengaruhi biaya asuransi, seperti kelompok usia, jenis kelamin, jumlah anak, wilayah tempat tinggal, dan status merokok.

---

**Tujuan:**

Mengidentifikasi faktor yang memberikan kontribusi biaya asuransi tertinggi serta membandingkan pengaruh masing-masing faktor terhadap variasi biaya asuransi.

---

**Visualisasi:**

Bar Chart (Grafik Batang).

```python
sns.set_style("whitegrid")

df['age_group'] = pd.cut(df['age'],
                         bins=[18, 30, 45, 60, 100],
                         labels=['18-30', '31-45', '46-60', '60+'])

factors = ['age_group', 'sex', 'children', 'region', 'smoker']
ranking = []

for col in factors:
    mean_values = df.groupby(col, observed=False)['charges'].mean()

    ranking.append({"factor": col.replace("_"," ").title(),
                    "avg_charges": mean_values.max()})

ranking_df = pd.DataFrame(ranking).sort_values(by="avg_charges",
                                               ascending=True)

plt.figure(figsize=(12,7))

sns.barplot(data=ranking_df,
            y="factor",
            x="avg_charges",
            hue="factor",
            palette="viridis",
            legend=False)

plt.title("Ranking Faktor dengan Kontribusi Biaya Asuransi Tertinggi", fontsize=16)
plt.xlabel("Rata-rata Charges")
plt.ylabel("Faktor")

plt.tight_layout()
plt.show()
```

Hasil perbandingan menunjukkan bahwa status merokok memiliki rata-rata biaya asuransi tertinggi dibandingkan faktor lainnya. Kelompok usia yang lebih tinggi juga menunjukkan biaya yang lebih besar, sedangkan jenis kelamin, jumlah anak, dan wilayah memiliki perbedaan biaya yang relatif kecil. Hal ini menunjukkan bahwa faktor perilaku dan usia merupakan determinan utama dalam variasi biaya asuransi.

## **Composition/Komposisi**

**Aktivitas:**

Menganalisis komposisi distribusi data berdasarkan seluruh faktor kategorikal yang mempengaruhi biaya asuransi, yaitu kelompok usia, jenis kelamin, jumlah anak, wilayah, dan status merokok.

---

**Tujuan:**

Mengetahui proporsi distribusi setiap kategori dalam dataset serta memahami karakteristik dominan populasi secara keseluruhan.

---

**Visualisasi:**

Pie Chart (Diagram Lingkaran).

```python
df['age_group'] = pd.cut(df['age'],
                         bins=[18, 30, 45, 60, 100],
                         labels=['18-30', '31-45', '46-60', '60+'])

factors = ['age_group', 'sex', 'children', 'region', 'smoker']

composition_list = []

for col in factors:
    temp = df[col].value_counts().reset_index()
    temp.columns = ['category', 'count']
    temp['factor'] = col
    temp['label'] = col + ": " + temp['category'].astype(str)
    composition_list.append(temp)

composition_df = pd.concat(composition_list)


threshold = composition_df['count'].sum() * 0.03

small = composition_df[composition_df['count'] < threshold]['count'].sum()
composition_df = composition_df[composition_df['count'] >= threshold]

if small > 0:
    composition_df = pd.concat([composition_df,
                                pd.DataFrame({'label':['Others'],
                                              'count':[small]})])

palette = sns.color_palette("tab10", len(composition_df))

plt.figure(figsize=(10,10))

wedges, texts, autotexts = plt.pie(composition_df['count'],
                                   labels=composition_df['label'],
                                   autopct='%1.1f%%',
                                   startangle=90,
                                   colors=plt.cm.Paired.colors,
                                   pctdistance=0.85)

plt.title('Komposisi Data Berdasarkan Berbagai Faktor', fontsize=16)
plt.tight_layout()
plt.show()
```

Grafik pie di atas menunjukkan bahwa beberapa kategori memiliki proporsi lebih dominan dibandingkan kategori lainnya, seperti kelompok usia tertentu dan kelompok non-perokok. Komposisi kategori yang beragam menunjukkan bahwa dataset mencerminkan berbagai karakteristik populasi yang berbeda.
